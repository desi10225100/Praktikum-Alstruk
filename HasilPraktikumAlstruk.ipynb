```python
# PRAKTIKUM 1 MENGENALI TIPE DATA
print("=== PRAKTIKUM 1 MENGEMAL TIPE DATA ===")
nama = "Fajar"
umur = 19
berat = 54.5
terdaftar = True

print(nama)
print(umur)
print(berat)
print(terdaftar)

print(type(nama))
print(type(umur))
print(type(berat))
print(type(terdaftar))

# Mengubah Nilai Umur Menjadi 20
umur +=1
print("umur baru: ", umur)
print("")



# PRAKTIKUM 2 OPERATOR ARITMATIKA
print("=== PRAKTIKUM 2 OPERATOR ARITMATIKA ===")
harga = 12500
jumlah = 4

total_harga = harga * jumlah
print("total_harga =", total_harga)
sisa_bagi = total_harga % 2000
print("sisa_bagi =", sisa_bagi)
harga_setelah_diskon = total_harga - 5000
print("harga_setelah_diskon =", harga_setelah_diskon)

# Menampilkan Hasil Setiap Perhitungan
print("Harga Satuan Buku = Rp.", harga)
print("Jumlah Buku Yang Dibeli = ", jumlah)
print("Total Harga = Rp.", total_harga)
print("Sisa Pembagian Total Harga Dengan 2000 = Rp.", sisa_bagi)
print("Harga Setelah Diskon = Rp.", harga_setelah_diskon)
print("")



# PRAKTIKUM 3 OPERATOR PERBANDINGAN DAN LOGIKA
print("=== PRAKTIKUM 3 OPERATOR PERBANDINGAN DAN LOGIKA ===")
nilai_mhs = 85
kehadiran_mhs = 90

syarat_nilai = nilai_mhs >= 75
print("Apakah memenuhi syarat?", syarat_nilai)
syarat_kehadiran = kehadiran_mhs >= 80
print("Apakah memenuhi syarat?", syarat_kehadiran)

memenuhi_kedua_syarat = syarat_nilai and syarat_kehadiran
print("LULUS PRAKTIKUM?", memenuhi_kedua_syarat)

# mengubah nilai menjadi 70
nilai_baru = 70

syarat_nilai = nilai_baru >= 75
print("Apakah memenuhi syarat?", syarat_nilai)
syarat_kehadiran = kehadiran_mhs >= 80
print("Apakah memenuhi syarat?", syarat_kehadiran)

memenuhi_kedua_syarat = syarat_nilai and syarat_kehadiran
print("LULUS PRAKTIKUM?", memenuhi_kedua_syarat)

# menggunakan operator or (nilai 70)
memenuhi_kedua_syarat = syarat_nilai or syarat_kehadiran
print("LULUS PRAKTIKUM?", memenuhi_kedua_syarat)
print("")


# PRAKTIKUM 4 OPERATOR PENUGASAN
print("=== PRAKTIKUM 4 OPERATOR PENUGASAN ===")
poin = 100
print("poin_awal :", poin)
# tanpa menambahkan poin lagi
# penjumlahan
poin +=25
print("poin1 : ", poin)
# pengurangan
poin -=10
print("poin2 : ", poin)
# perkalian
poin *=2
print("poin3 : ", poin)
# pembagian
poin /=2
print("poin4 : ", poin)

# menambahkan poin lagi
print("== menambahkan poin lagi ==")
# penjumlahan
poin = 100
poin +=25
print("poin1 : ", poin)
# pengurangan
poin = 100
poin -=10
print("poin2 : ", poin)
# perkalian
poin = 100
poin *=2
print("poin3 : ", poin)
# pembagian
poin = 100
poin /=2
print("poin4 : ", poin)
print("")


# PRAKTIKUM 5 STRUKTUR DATA DASAR
print("=== PRAKTIKUM 5 STRUKTUR DATA DASAR ===")
nilai = [78, 85, 90, 85]
mahasiswa = {"nama": "Nadia", "kelas": "IF-A"}
koordinat = (5, 8)
angka = {2, 4, 4, 6, 8}

# menampilkan nilai pertama dari list
print("Nilai pertama: ", nilai[0])

# menambahkan nilai 95 ke list dan menampilkan list terbaru
nilai.append(95)
print("Nilai terbaru list: ", nilai)

# menampilkan nama mahasiswa
print("Nama Mahasiswa: ", mahasiswa["nama"])

# menampilkan koordinar pertama dari tupple
print("Koordinat pertama dari tupple: ", koordinat[0])

# menampilkan angka unik dengan sorted(angka)
print("Angka unik (sorted): ", sorted(angka))
print("")

# TANTANGAN TERINTEGRASI (PROGRAM REKAP NILAI)
nama = "DESI SRI ANJANI"
nilai_tugas = 80
nilai_ujian = 72

# rata-rata niali
rata_rata = (nilai_tugas + nilai_ujian) / 2
# status lulus (rata-rata >= 75)
status_lulus = rata_rata >= 75

print("==== TANTANGAN TERINTEGRASI ====")
print("=== Nilai Mahasiswa ===")
print("Nama Mahasiswa: ", nama)
print("Nilai Tugas: ", nilai_tugas)
print("Nilai Ujian: ", nilai_ujian)
print("Rata-rata: ", rata_rata)
print("Status Lulus: ", status_lulus)

# uji program dengan nilai lain
nilai_tugas2 = 70
nilai_ujian2 = 78

# rata-rata niali
rata_rata2 = (nilai_tugas2 + nilai_ujian2) / 2
# status lulus (rata-rata >= 75)
status_lulus2 = rata_rata2 >= 75

print("=== Nilai Mahasiswa 2 ===")
print("Nama Mahasiswa: ", nama)
print("Nilai Tugas: ", nilai_tugas2)
print("Nilai Ujian: ", nilai_ujian2)
print("Rata-rata: ", rata_rata2)
print("Status Lulus: ", status_lulus2)
```
